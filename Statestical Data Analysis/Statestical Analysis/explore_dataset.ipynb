```python
# import pandas libraries
import pandas as pd
import numpy as np

# ignore warning
import warnings
warnings.filterwarnings('ignore')
```

```python
# read dataset from sales_transaction_dataset.csv
df = pd.read_excel('../dataset/Lab_1.0_Baseball_Salaries.xlsx')
```

```python
# show top 5 records
df.head()
```

```python
df.plot()
```

```python
[val_mean,val_median,val_mode] = [df['Salary'].mean(),df['Salary'].median(),df['Salary'].mode()]
print(val_mean)
print(val_median)
print(val_mode)
```

```python
# find min max
[min, max] = [df['Salary'].min(), df['Salary'].max()]

print(min)
print(max)
```
